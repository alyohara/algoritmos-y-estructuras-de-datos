# Erathostene's sieve

The sieve of Erathostene is a relatively efficient way to generate all prime numbers below a number $N$ without making any division. The algorithm is the following :

- Begin with a list $L$ of numbers from $1$ to $N$ and an empty list $P$ of prime numbers.
- Cross $1$ in it. 
- While there is still an uncrossed number in the list $L$, take the smallest uncrossed number $p=L[p-1]$ in $L$, add it to $P$ and cross it in $L$.  Cross every number in $L$ of index $p-1+k p$ where $k$ is a positive integer.
- Return $P$.


Implement this algorithm to compute the list of  prime numbers between $2$ and $10^6$. 

Using the list of prime numbers (using pyplot) to  illustrate the validity of the Hadamard-de la Vallée-Poussin (a.k.a. the prime numbers theorem) : if $\pi(x)$ is the number of prime numbers between $1$ and $x$ then :

$$\frac{x}{\pi(x)}\sim_{x\to \infty} \log(x) $$

# Asymptotic behavior.

Let $f:\mathbb{R}\to\mathbb{R}$, $x_0\in\mathbb{R}$. A sequence  $(f^n(x_0))_n$ is eventually periodic at rank $N_0$ with orbit $[y_1,\dots,y_k]$ if  $y_1,\dots,y_k$ are mutually distinct points such that $f(y_i)=y_{i+1}$ for $1\leq i\leq k-1$, $f(y_k)=f(y_1)$ and such that there exists $N_0$ verifying  $f^{N_0}(x_0)=y_1$, $f(y_i)=y_{i+1}$ for $1\leq i\leq k-1$. Remark that if $k=1$ then $f$ is eventually constant. 


Fix an integer $limrank$. For a given $f:\mathbb{R}\to\mathbb{R}$ and $x_0\in\mathbb{R}$, the asymptotic behavior of the sequence $(f^n(x_0))$ is 

- its orbit if it is eventually periodic at a rank smaller than $limrank$,

- the list with one element : $[f^{limrank}(x_0)]$ else. 

Write a function that, given three arguments : a map $f$, a real number $x_0$ and an integer $limrank$ computes the asymptotic behavior of the sequence as above.
