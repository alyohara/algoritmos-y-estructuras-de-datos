**Cifrado César:** https://es.wikipedia.org/wiki/Cifrado_C%C3%A9sar


Sea $\mathcal{A}$ un conjunto finito de caracteres, el cual será nuestro **alfabeto**. Sea $N$ la carinalidad de  $\mathcal{A}$. Asumiremos que existe una función $f$ la cual asigna un número unico a una letra. 

La **Llave** $K$ del cifrado es simplemente algún caracter para transformar nuestro string:

$$s(0)s(1)\dots s(\ell)\text{ to } f^{-1}(f(s(0))+f(K))\dots  f^{-1}(f(s(\ell))+f(K)) $$


Es decir, *'desplazar cada caracter del string $K$ lugares'*. Podemos utilizar la operación *'modulo'* de $N$ para darle sentido al despazamiento. Llamamos $\mathcal{C}_K$ a la función de cifrado asociada a $K$. 



Ejemplo, si $\mathcal{A}=[A,\dots,Z]$, la función $f$ puede ser $f(\text{una letra})=$ posición en el alfabeto. Entonces, tomando la llave $B$ ($f(B)=1$):

$$\mathcal{C}_B('APPLES')='BQQMFT' $$

$$\mathcal{C}_B('ZOO')='APP' $$



**Podemos utilizar las funciones `chr` y `ord` para pasar de caracteres a números y vice-versa.**
