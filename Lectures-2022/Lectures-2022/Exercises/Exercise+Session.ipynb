We denote Sn the set of permutations of {0, ... , n − 1} . By definition, the composition of σ1 and σ2 in Sn is a permutation denoted σ1 ∘ σ2 sending i in {0, ... , n − 1} to σ1 (σ2 (i)). See your undergraduate course on group theory for an exposition on this or https://en.wikipedia.org/wiki/Symmetric_group (https://en.wikipedia.org/wiki/Symmetric_group).


In the sequel, a permutation σ will be represented by the tuple (σ(0), ... , σ(n − 1)).

The Hanoi towers is a  combinatorial problem. You have three  sticks (the towers) and $N$  disks of different radii (radiuses?) with a hole at their centers so that you can insert a disk around a stick. At the beginning of the problem, the $N$ disks are put in decreasing order of radius (the biggest disk is at the bottom)  on one single stick. You need to put the $N$ disks in decreasing order of radius around another stick minimizing your moves. 

A **move** in this problem consists of taking a disk on top of one of the tower and putting on top of another tower (the **rule** needs to be checked though). 

**The main rule of the problem is that, on any tower, at any time, if one disk of radius $R_1$ is above another one of radius $R_2$ then $R_1<R_2$. **

### Cesar cipher, algorithm

Let us say you have a finite set of characters $\mathcal{A}$ which will be the **alphabet**. Let $N$ be the cardinal of $\mathcal{A}$. Assume that you have a map $f$ that univoquly turns a character to a number.  

The **key** $K$ to this cipher is simply some character then you transform the string of characters :

$$s(0)s(1)\dots s(\ell)\text{ to } f^{-1}(f(s(0))+f(K))\dots  f^{-1}(f(s(\ell))+f(K)) $$


In other words, you **shift every character of the string by the same number associated to $K$**. Of course, you are supposed to work modulo $N$ for the order to make sense in any case. We denote $\mathcal{C}_K$ the coding function associated to $K$. 


Since it is obvious that $\mathcal{C}_K\circ\mathcal{C}_{-K}$ is the identity, this is a legitimate cipher.  


As a baby example, if you take $\mathcal{A}=[A,\dots,Z]$, the map $f$ can simply be defined $f(\text{some letter})=$ its rank in the alphabet. Then, taking the key $B$ ($f(B)=1$) then :

$$\mathcal{C}_B('APPLES')='BQQMFT' $$

$$\mathcal{C}_B('ZOO')='APP' $$

### The python script for Cesar cipher

Here we take $N=1000$. Of course the alphabet will be the first $1000$ characters in utf-8 and we will use the built-in python functions $\textbf{chr}$ and $\textbf{ord}$ to go from numbers to characters and conversely from characters to numbers.

```python
N=1000


def Cesarcipher(string,K):
    
    #Turn the string of characters to a list of their associated numbers
    L=list(map(lambda x : ord(x),list(string))) 
    
    #Shift every number of the list with the order of K/work modulo N
    L=list(map(lambda x : (x+ord(K))% N,L)) 
    
    #Turn back the 'crypted' numbers to a regular list of characters
    L=list(map(lambda x : chr(x),L)) 
    
    #Construct the crypted string out of the list of crypted characters
    res=''
    for x in L:
        res+=x
        
        
    return res
    
```

```python
import random

#Take a random key
K=chr(random.randrange(0,N))

#Compute the decrypting key
Decryptkey=chr(N-ord(K))

stringtoencrypt='I want to cipher this string'

#Cipher
encrypted=Cesarcipher(stringtoencrypt,K)

#Decipher
decrypted=Cesarcipher(encrypted,Decryptkey)


print("Original message : ",stringtoencrypt)
print("Crypted message  : ",encrypted)
print("Decrypted message : ",decrypted)
```

Furthermore, you can safely assume that the average length of a word in an English text is below 15. Therefore if the proportion of spaces among the whole set of characters in your string is below $1/15$ then your string of characters is probably not an English text.

```python
def IsEnglishtext(s):
    #number of spaces(initalized to 1 to ensure no division by zero)
    nspaces=1
    
    #number of 'nonregular' characters(initalized to 1 to ensure no division by zero)
    nnonreg=1
    
    for c in s:
        if c==' ':
            nspaces+=1
        if ord(c)>130 and ord(c)<32:
            nnonreg+=1
    return len(s)//nspaces<=15 and len(s)//nnonreg>=10
```

```python
#I copy past the triangle class 
from turtle import *
import math
#A map that check if two tuples are colinear
arecolinear=lambda u,v: u[1]*v[0]==u[0]*v[1] 

#A map that computes the norm of a vector
normvect=lambda u: math.sqrt(u[0]**2+u[1]**2)


class triangle():
    
    def __init__(self,content):
        if type(content)!=type((0,1,2)):
            raise TypeError("The content should be a triplet of points")
        elif len(content)!=3:
            raise ValueError("There should be three points in the content")
        else:  
            self.content=content
    
    def isflat(self):
        A,B,C=self.content #Define A,B,C to be the points defining your triangle
        u=(B[0]-A[0],B[1]-A[1])  #u=the vector AB
        v=(C[0]-A[0],C[1]-A[1])  #v=the vector AC
        return arecolinear(u,v) 
    
    def perimeter(self):
        A,B,C=self.content #Define A,B,C to be the points defining your triangle
        u=(B[0]-A[0],B[1]-A[1]) #u=the vector AB
        v=(C[0]-A[0],C[1]-A[1]) #v=the vector AC
        w=(C[0]-B[0],C[1]-B[1]) #w=the vector BC
        return normvect(u)+normvect(v)+normvect(w)
    
    
    #To compute the area of a triangle see the picture above 
    
    def area(self):
        A,B,C=self.content #Define A,B,C to be the points defining your triangle
        u=(B[0]-A[0],B[1]-A[1])  #u=the vector AB
        v=(C[0]-A[0],C[1]-A[1])  #v=the vector AC
        scalarprod_uv=u[0]*v[0]+u[1]*v[1]
        h=math.sqrt(normvect(u)**2-scalarprod_uv**2/normvect(v)**2)
        return normvect(v)*h/2
    
    
    def is_trianglerectangle(self):
        A,B,C=self.content  #Define A,B,C to be the points defining your triangle
        u=(B[0]-A[0],B[1]-A[1]) #u=the vector AB
        v=(C[0]-A[0],C[1]-A[1]) #v=the vector AC
        w=(C[0]-B[0],C[1]-B[1]) #w=the vector BC
        if u>=v and u>=w:
            return v[0]**2+v[1]**2+w[0]**2+w[1]**2==u[0]**2+u[1]**2
        if v>=u and v>=w:
            return u[0]**2+u[1]**2+w[0]**2+w[1]**2==v[0]**2+v[1]**2
        else:
            return v[0]**2+v[1]**2+u[0]**2+u[1]**2==w[0]**2+w[1]**2
        
        
    def draw(self,tcolor=None,coord=False):
        A,B,C=self.content  #Define A,B,C to be the points defining your triangle
        
        if tcolor==None: #If I don't ask for color
            penup() #do not draw
            setposition(A) #Go to the point A
            pendown() #draw 
            setposition(B) #go to the point B (from A to B) and thus draw AB
            setposition(C) #go to the point C (from B to C) and thus draw BC
            setposition(A) #go to the point A (from C to A) and thus draw CA
            penup() #stop drawing
            
            
        else: #Meaning that I entered a color in tcolor
            fillcolor(tcolor)  #Set the color to fill to tcolor
            penup() #do not draw
            setposition(A) #Go to the point A
            pendown() #draw           
            begin_fill() #From there to end_fill the inside of every figure drawn will be filled with tcolor
            setposition(B) #go to the point B (from A to B) and thus draw AB
            setposition(C) #go to the point C (from B to C) and thus draw BC
            setposition(A) #go to the point A (from C to A) and thus draw CA
            end_fill() #Stop filling
            penup() #stop drawing
            
            
        if coord: #if coord=True then it prints the coordinates on each point. The pen needs to be up.
            setposition(A)
            write("({},{})".format(A[0],A[1]), True, align="center",font=( "Arial", 16, "normal") )
            setposition(B)
            write("({},{})".format(B[0],B[1]), True, align="center",font=( "Arial", 16, "normal") )
            setposition(C)
            write("({},{})".format(C[0],C[1]), True, align="center",font=( "Arial", 16, "normal") )
                

            
    
```

# Erathostene's sieve

The sieve of Erathostene is a relatively efficient way to generate all prime numbers below a number $N$ without making any division. The algorithm is the following :

- Begin with a list $L$ of numbers from $1$ to $N$ and an empty list $P$ of prime numbers.
- Cross $1$ in it. 
- While there is still an uncrossed number in the list $L$, take the smallest uncrossed number $p=L[p-1]$ in $L$, add it to $P$ and cross it in $L$.  Cross every number in $L$ of index $p-1+k p$ where $k$ is a positive integer.
- Return $P$.
