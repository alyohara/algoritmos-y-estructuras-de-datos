$ a_n = a_1 + \sum_{n=1}^{n-1} 2n$

($n >= 2$ y $a_1 = 0$)

#### Solución

La serie que genera dicha lista es:

$ a_n = a_1 + \sum_{n=1}^{n-1} 2n$

($n >= 2$ y $a_1 = 0$)

Podemos escribir la serie $a_n$ como:

$\begin{align}
a_n & = a_1 + \sum_{n=1}^{n-1} b_n\\
& = 0 + 2 \cdot \frac{(n-1) \cdot n}{2}\\
& = n^2 - n\\
\end{align}$

### Ej. 8:


**Cifrado César:** https://es.wikipedia.org/wiki/Cifrado_C%C3%A9sar


Sea $\mathcal{A}$ un conjunto finito de caracteres, el cual será nuestro **alfabeto**. Sea $N$ la carinalidad de  $\mathcal{A}$. Asumiremos que existe una función $f$ la cual asigna un número unico a una letra. 

La **Llave** $K$ del cifrado es simplemente algún caracter para transformar nuestro string:

$$s(0)s(1)\dots s(\ell)\text{ to } f^{-1}(f(s(0))+f(K))\dots  f^{-1}(f(s(\ell))+f(K)) $$


Es decir, *'desplazar cada caracter del string $K$ lugares'*. Podemos utilizar la operación *'modulo'* de $N$ para darle sentido al despazamiento. Llamamos $\mathcal{C}_K$ a la función de cifrado asociada a $K$. 



Ejemplo, si $\mathcal{A}=[A,\dots,Z]$, la función $f$ puede ser $f(\text{una letra})=$ posición en el alfabeto. Entonces, tomando la llave $B$ ($f(B)=1$):

$$\mathcal{C}_B('APPLES')='BQQMFT' $$

$$\mathcal{C}_B('ZOO')='APP' $$



**Podemos utilizar las funciones `chr` y `ord` para pasar de caracteres a números y vice-versa.**

#### Solución

```python
# Si bien esta solucion es mas compleja, es posible resolver el ejercicio con ciclos simples while/for.

# Con N=1000 generamos un alfabeto con los primeros 1000 caracteres utf-8.
N=1000


def Cesarcipher(string,K):
    
    # Convierte el string en su equivalente numerico
    L=list(map(lambda x : ord(x),list(string))) 
    
    # Dezplazar la lista
    L=list(map(lambda x : (x+ord(K))% N,L)) 
    
    # Convierte a caracteres nuevamente
    L=list(map(lambda x : chr(x),L)) 
    
    # Contruir el mensaje cifrado, utilizando los números cifrados 
    res=''
    for x in L:
        res+=x
        
        
    return res
    
    
import random

#Take a random key
K=chr(random.randrange(0,N))

#Compute the decrypting key
Decryptkey=chr(N-ord(K))

stringtoencrypt='I want to cipher this string'

#Cipher
encrypted=Cesarcipher(stringtoencrypt,K)

#Decipher
decrypted=Cesarcipher(encrypted,Decryptkey)


print("Original message : ",stringtoencrypt)
print("Crypted message  : ",encrypted)
print("Decrypted message : ",decrypted)
```
