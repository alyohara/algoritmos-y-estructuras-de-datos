## What is $\LaTeX$?

$\LaTeX$ is a langage to write (scientific) documents. It is also the software converting the said langage into readable documents (usually .pdf portable document format or .ps postscript). 

For instance, the langage $\LaTeX$ is used in the Python notebook to write some mathematics, here is the binomal formula  :

## What is  $\LaTeX$ useful for ?

$\LaTeX$ is used by many scientists to write **articles, reports and books**. Its most interesting feature being at first to enable you to write mathematical formula in a very nice way. ## How to write plain text (text mode)

When you are writing English sentences in your text, the text you get after compilation is very closed to what you are writing.

However, what is happening if I am doing one single line break?

Remark                                      that                                      if                                      you                                      have                                      long                                     spaces                                      then                                      you                                      end                                      up                               with                                      small                                      spaces                                      after                                      compilation.

The  $\LaTeX$ langage is meant to help you writing and read your **script**. Indeed, by doing so, you can use line breaks to make a more readable script without messing with the final output of your work (in math mode line breaks and spaces are almost non-significant). 

This is a very good thing because, believe me, you cannot make a nice effective script in one shot. You will get compilation errors, grammatical mistakes, and so on. You really want to take care about this and you cannot do it nicely if your script is not readable. 

Should you want to break line in the output, it suffices to break the line twice in your script. 

Here it is.

There are few symbols that have a specific meaning in $\LaTeX$, you cannot use them in plain text. 


- Any piece of mathematic should be put between two dollar signs '$\$$' because this is the way you make $\LaTeX$ aware that you want to be in math mode.

- Any command that is not directly available on your keyboard should be specified with a keyword written right after a backslash '$\backslash$'. 

- Any line of command written after a percentage sign  '$\%$' is neutralized (you actually are making comments in your $\LaTeX$ script like you would use a $\#$ in Python). 

- The hashtag $\#$ is used to specify the number of an argument when defining command.

- The esperluette $\&$ is used to specify alignment in equations and tabular.

- The hat '^' is used to specify an exponent. 

- The underscore '_' is to specify an index.

- Finally the opening and closing braces '$\{$' and '$\}$' are usually used to specify some sub-environments or arguments for commands.

These characters should always be used carefully in a $\LaTeX$ script because when the program reads them in a script, it does not print them as characters but is looking for a specific order to execute. 

## How to write mathematics (math mode)

The $\LaTeX$ langage is all about inserting 'smoothly' mathematics inside an English text. To do so, you need to switch from the text mode to the math mode. The straightforward way to do it is to use dollar signs.

Can you tell the difference between $2+3=5$ and 2+3=5 ?

Mathematics in formulas are slightly oversized, the spacing is improved. Furthermore, like a text editor would do it will cut very long formulas to make sure it does not go further than the  right end of the page. In other words it takes care of the things you don't want to take care of. For instance $1+2+3+4+5+6+7+8=36$. I did not break the line by myself, it did it automatically.

Some symbols are written on your keyboard and appear in math mode as they appear on your keyboard. For instance $*$, $+$, $=$, $($, $)$  and you can write as is in **math mode**.

Most printable symbols are not on your keyboard though. In order to print a character, you first need to know its name and then to write '$\backslash$nameofthesymbol' in math mode. Here is a (non-exhaustive) list of examples :

- $\backslash$gamma : $\gamma$
- $\backslash$Gamma : $\Gamma$
- $\backslash$backslash : $\backslash$
- $\backslash$forall : $\forall$
- $\backslash$exists : $\exists$
- $\backslash$in : $\in$
- $\backslash$neq : $\neq$
- $\backslash$mathbb{N} : $\mathbb{N}$
- $\backslash$| : $\|$
- $\backslash$frac{a}{b} : $\frac{a}{b}$
- $\backslash$sqrt[3]{2} :  $\sqrt[3]{2}$

** Example ** : Write, with $\LaTeX$ the definition for the  continuity of a function $f$ from $\mathbb{R}$ to $\mathbb{R}$ and for the uniform continuity of a function $f$ from $\mathbb{R}$ to $\mathbb{R}$

The function $f$ is continuous on $\mathbb{R}$ if $\forall x\in \mathbb{R}$, $\forall \epsilon >0$, $\exists \delta>0$  such that $\forall y\in\mathbb{R}$ verifying $|x-y|<\delta\Rightarrow |f(x)-f(y)|<\epsilon$.

The function $f$ is uniformly continuous on $\mathbb{R}$ if $\forall \epsilon >0$, $\exists \delta>0$  such that $\forall x,y\in\mathbb{R}$ verifying $|x-y|<\delta\Rightarrow |f(x)-f(y)|<\epsilon$.

The convention is that characters used to represent the names of variables (therefore written in math mode) should always be written differently from characters used to represent words in a sentence. This is why the 'plain text' in math mode behaves badly : it is not meant to represent plain text.

## Good and bad practice when you write with $\LaTeX$

- Even though, you are writing a scientific document in $\LaTeX$, you  want to check grammar and spelling (not always easy, I know). Usually text editors come with a nice spell checker, you might want to use it. 

- The ponctuation is important too.


- When you first discover $\LaTeX$, especially when it gets easier (you remember the names of the commands) you want to put math mode everywhere. Remember that the same rule you have to respect when you answer a question in your exam applies when you write a report in science. You need to avoid mixing mathematical symbols with English sentences. 

- Use books of mathematics you like and try to write the same things and the same way  as they do.

- It is usually a good idea to really try to play the role of the (critical) reader after you have written your pdf file. 

- Some symbols should be written in a way and not another way. This is a convention, even though people might understand it, they won't like it. For instance if you want to write a bra-ket $|\psi>$ you should use $|\psi\rangle$.

- Math operators such as sinus, cosinus, determinant and many others should not be written like this $sin(\frac{\pi}{2})$. They should be written like this instead : $\sin(\frac{\pi}{2})$.

## The body of the document

In the script for your $\LaTeX$ file, the body of the document (what will actually be output in the pdf) is written between 

$\backslash$ begin{document} and   $\backslash$end{document}

For instance, this is where you should be writing plain text and math mode like we have seen before. There are  a few things that seem very important to me to give a nice structure to your document. 

### sections, subsections, theorems, label and references, proof environment


By using \section{title of the section}, \subsection{title of the subsection}, you  create different parts in your article. The numeration is automatic. Furthermore you can call \tableofcontents to actually write down the whole table of contents enumerating all sections, subsections. 

If you want to write a 'Lemme' then you should write 

$\backslash$begin{lemme}[Name of the lemme (optional)]

$\dots$

$\backslash$end{lemme}

Remark that you need to have defined what is a 'lemme' with \newtheorem{lemme}{Lemme}. 

One nice feature with these commands is that they come with automatic numeration. For instance, if you have forgotten to write some lemma at the beginning of your article, you don't have to change the numeration of lemmas afterward, it will be automatically changed.

Furthermore, if you want to use the number of your 'lemme', for instance : "in Lemma 3, I proved that an apple a day keeps the doctor away", you should not write the number by yourself (it might change at some point). Instead, you should do the following :

$\backslash$begin{lemme}[Name of the lemme (optional)]\label{thelemmaIwilluse}

An apple a day keeps the doctor away.

$\backslash$end{lemme}

And then "in Lemma $\backslash$ref{thelemmaIwilluse}, I proved that an apple a day keeps the doctor away". 

After a proposition, theorem, lemma, you usually have a proof for it. By writing :

$\backslash$begin{proof}

proof


$\backslash$end{proof}

you have a nice environment that explicitely emphasize to the reader where begins the proof and where it ends.

## How to handle errors in compilations

Like in Python, you should not be afraid about having fatal errors or errors in your script. It comes with the learning. When $\LaTeX$  compiles a script, it creates a bunch of auxiliary files (helping with the references, the macro you are using) one of them is a .log file. It contains a log of what happened during the compilation of your file. 

By scrolling through this log, you will spot errors. Usually, the text editor directly helps you with the errors. The errors are always given a line number, this is the line number in the script. It will help you to directly go into your script and spot the error you made. 

If you cannot figure out what is wrong (usually a mispelled command, also check out if you have closed every environment you opened) try to retype the thing or copy/past it from where you have found it.
