In this homework, we assume that we only have  disks of at most $10$ different lengths in our Hanoi Towers (that should help us to have a good scale for the different disks). 



Q1. Add a method **draw** having two extra arguments to the definition of this class such that by calling $HT.\textbf{draw}(x_0)$, you  draw (using **turtle**) the HanoiTower $HT$ with the $x$-coordinate centered at $x_0$. Make sure that one can easily distinguish one disk from the other. 

Q2. Write a script drawing all steps of the resolution of the Hanoi Towers. 

```python
#Q1.

import random 
from  turtle import *
N=20      #Number of disks overall
colors=[(0,n/N,1-n/N) for n in range(0,N)] #All the colors

class HanoiT():


    def __init__(self,content): # Initialization, just the content
                                # Warning : using mutable keyword arguments is bad practice
        self.content=content
        

    
    def __str__(self):    # It is always good to give python a way to print objects
        return str(self.content)
    
    def destroy(self):  # Destroy the top element of self, raise an error if no elements
        del self.content[-1]
    
    
    def adder(self,t):   # Adding t to the top of self
        if len(self.content)>0 and t>= self.content[len(self.content)-1]: # Check the condition for the HanoiT
            raise ValueError("The disk you are adding is bigger than the top of the Hanoi Tower")
        else:
            self.content.append(t)
   
    #--------# ^copy-past from the correction of last homework^


 # The base of the Hanoi tower is of length 125 centered at x0 and of y-coordinate -250.
 # The height of the Hanoi tower is of 550 
 # The biggest disk should be of length 100
 # The height of a disk should be 500/N
 # Therefore the disk labelled n should be of length 100*n/N
 # The color of the n-th disk is colors[n-1]

    def draw(self,x0):
        speed(0)
        penup()
        seth(0)
        setposition(x0-62.5,-250)
        pendown()
        forward(125)
        penup()
        setposition(x0,-250)
        left(90)#Turn
        pendown()
        forward(500)
        
        #Height of a disk
        heightdisk=500/N
        y=-250
        for n in self.content:
            #Set the width of the disk
            width=100*n/N
            penup()
            seth(0)
            setposition(x0-width/2,y)
            fillcolor(colors[n-1])
            pendown()
            begin_fill()
            forward(width)
            left(90)
            forward(heightdisk)
            left(90)
            forward(width)
            left(90)
            forward(heightdisk)
            left(90)
            end_fill()
            y+=heightdisk



        
```

```python
#Test
reset()
HT=HanoiT([4,3])
HT.draw(0)
```

```python
#Q2

tcounter=0


def move(T1 : HanoiT, T2 : HanoiT): #It does not check the type for you, 
                                    #but is printed when you call help(move)
    global tcounter
    if len(T1.content)==0:
        raise ValueError("T1 has no disk")
    else:
        tcounter+=1
        top=T1.content[len(T1.content)-1]
        T1.destroy()
        T2.adder(top)
        


# At each step, you have at most 3 possible moves
# Out of these moves, one of them is the reverse of something you have just done
# So you basically have only two interesting moves 
# Solving this puzzle is basically choosing the good move at each step

# The initialization is tricky, since the best move to make 
#at the beginning (you only have two possibilities)
# depends on the parity of N. 

# Remark that if you move a disk twice in a row, this is not a good sequence of moves 
# (you could have done it in one move)
reset()
hideturtle() 



HT1=HanoiT([N-i for i in range (0,N)])
HT2=HanoiT([])
HT3=HanoiT([])

small=False # A boolean saying if you have just moved the smallest disk or not.
lastmove=HT1 #  the last Hanoi tower where you added an element
# Auxiliary list is decreasingly listing  the Hanoi Towers according to the size of top disk
# By definition an empty list has an infinitely big top disk
Auxlist=[3,2,1]
def Ordering(): # Will return an Auxlist according to the top element
    if len(HT[0].content)==0:
        t1=N+3
    else:
        t1=HT[0].content[-1]
    if len(HT[1].content)==0:
        t2=N+2
    else:
        t2=HT[1].content[-1]
    if len(HT[2].content)==0:
        t3=N+1
    else:
        t3=HT[2].content[-1]
        
    if t1<t2:
        if t2<t3:
            return([3,2,1])
        elif t3<t1: # and t1<t2
            return([2,1,3])
        else:#t1<t2 and t1<t3 and t3<t2
            return([2,3,1])
    else:#t2<t1
        if t1<t3:
            return([3,1,2])
        elif t3<t2:
            return([1,2,3])
        else:
            return([1,3,2])

T=0     #The bigger it is the more time you have between two frames drawing the three Hanoi Towers

def Htowers(HT1,HT2,HT3,T):
    #We want to draw each step at once.
    tracer(1000)
    clear()
    
    #Draw the first HanoiT
    HT1.draw(-200)
    penup()
    setposition(-200,-275)
    pendown()
    write("Hanoi Tower 1",align="center",font=("Arial", 15, "normal"))
    
    #Draw the second HanoiT
    HT2.draw(0)
    penup()
    setposition(0,-275)
    pendown()
    write("Hanoi Tower 2",align="center",font=("Arial", 15, "normal"))
    
    #Draw the third HanoiT
    HT3.draw(200)
    penup()
    setposition(200,-275)
    pendown()
    write("Hanoi Tower 3",align="center",font=("Arial", 15, "normal"))
    
    #Draw a counter
    penup()
    setposition(0,320)
    pendown()
    write("Number of steps : {}".format(tcounter),align="center",font=("Arial", 15, "normal"))
    tracer(1)
    
    #Here I am waiting by just rotating the turtle by an angle +time then -time. 
    speed(1)
    left(T)
    right(T)
    speed(0)

    

# List of Hanoi Towers, it helps to have them as a list of Hanoi towers 
#rather than as three Hanoi towers.  
HT=[HT1,HT2,HT3]


# initialization, it depends on the parity of N
if N%2==1:
    move(HT1,HT3)
    Auxlist=[2,1,3]
    lastmove=3
else:
    move(HT1,HT2)
    Auxlist=[3,1,2]
    lastmove=2
small=True



while tcounter<10000 and len(HT3.content)<N:
    Htowers(HT1,HT2,HT3,T) #Draw the towers
    if not(small): # If you did not move the smallest element in the last move, do it now.
        #The hardest part of the problem is to understand 
        #where you should put the smallest element 
        #because there are 2 choices there
        if (HT[Auxlist[2]-1].content[-1]-HT[Auxlist[1]-1].content[-1])%2==0: 
            move(HT[Auxlist[2]-1],HT[Auxlist[0]-1])
        else:
            move(HT[Auxlist[2]-1],HT[Auxlist[1]-1])
        Auxlist=Ordering()
        small=True # You just changed the smallest element
    else: #If you just changed the smallest element, you only have one choice
          #, to move the second smallest element 
          #to the HanoiTower with the biggest element
        move(HT[Auxlist[1]-1],HT[Auxlist[0]-1])
        Auxlist=Ordering()
        small=False # You did not change the smallest element
Htowers(HT1,HT2,HT3,T) #Draw the towers
```

# Open and close a file 
Using Python, you can automatically process text files. By processing I mean either **read** a file or **write** on it. To do this, you need to $\textbf{open}$ a file. 

$\textbf{open}(path, string)$

The first argument is the path pointing toward the file that you want to open.  The second argument is an optional argument which specifies what you want to do with your file. $'r'$ means that you want to read it, $'w'$ means that you will create a file with the given path or rewrite on an existing file anew, $'r+'$ means that you will read it and write on it, $'a'$ means that you will write at the end of the existing file. 

You may also add a 'b' to the second argument meaning that you will do the specified action but considering your file as a binary file rather than a text file. 

```python
p=Path.home()/'Desktop'
f=open(p/'Exemple/old.py', 'r') #This creates a variable f which "contains" the specified file
```

```python
print(type(f))
```

```python
print(f.read())#You can do many things with your file this is what we are going to see next
```

```python
f.close()# close the file
f.closed # Ask if the file is closed
```

```python
f.read()
```

Once a file $f$ is closed, $f$ is still assigned but you cannot use the file anymore, to do it, you need to reopen it again. The reason is clear, while the file $f$ is closed it only uses as much memory as needed to specify the path to the file but when the file is open the memory you use is as big as the file is. In particular, you will have an Error message if you attempt to open a file which is twice bigger as what your session is authorized to use. 

Let us say that leaving open many files is a very bad practice since it will have a bad influence on the efficiency of your program. The Python documentation highlights a specific command to make sure that you only open a file when you need it. The command is $\textbf{with}$. 

```python
#instead of  'Exemple/old.py' take any file in some directory of your desktop#
with open(p/'Exemple/old.py', 'r') as f:
    print(f.read()) #Do whatever you want with your file : the file is open
f.closed #then the file is closed
#As we can see, the file is only open in the indented block. 
```
